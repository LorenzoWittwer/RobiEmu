```python
# %%writefile GUI.py

# Bibliotheken 
import customtkinter as ctk

# Projektmodule
from ROBI import ROBI

'''

'''
class GUI(ctk.CTk):

    def __init__(self):
        super().__init__()

        self.title('RobiEmu von Lorenzo Wittwer')
        self.geometry('1400x850')
        self.resizable(False, False)

        self.create_window() # GUI erstellen

    '''

    '''
    def create_window(self):
        # Fensterrahmen 
        self.window_frame = ctk.CTkFrame(self, fg_color='#1a1933')
        self.window_frame.pack(fill='both', expand=True)

        # Spalten
        self.window_frame.grid_columnconfigure(0, weight=0)
        self.window_frame.grid_columnconfigure(1, weight=1)
        self.window_frame.grid_columnconfigure(2, weight=0)

        # Zeilen
        self.window_frame.grid_rowconfigure(0, weight=1)
        self.window_frame.grid_rowconfigure(1, weight=0)

        # Panels im Fenster erstellen
        self.create_leftPanel()
        self.create_rightPanel()

        # Simulationsbereich im Fenster erstellen
        self.create_simulationArea()

        # Untere Leiste erstellen
        self.create_bottomBar()

    '''
    
    '''
    def create_leftPanel(self):
        # Rahmen 
        self.leftPanel_frame = ctk.CTkFrame(self.window_frame, width=300, fg_color='#1a1933')
        self.leftPanel_frame.grid(row=0, column=0, sticky='nsew')
        self.leftPanel_frame.grid_propagate(False)

    '''
    
    '''
    def create_rightPanel(self):
        # Rahmen 
        self.rightPanel_frame = ctk.CTkFrame(self.window_frame, width=300, fg_color='#1a1933')
        self.rightPanel_frame.grid(row=0, column=2, sticky='nsew')
        self.rightPanel_frame.grid_propagate(False)
        
    '''
    
    '''
    def create_simulationArea(self):
        # Rahmen 
        self.simulationArea_frame = ctk.CTkFrame(self.window_frame, fg_color='white')
        self.simulationArea_frame.grid(row=0, column=1, sticky='nsew')

        # Spalte
        self.simulationArea_frame.grid_columnconfigure(0, weight=1)

        # Zeile
        self.simulationArea_frame.grid_rowconfigure(0, weight=1)

        self.robi = ROBI(self.simulationArea_frame)
        self.robi.grid(row=0, column=0)

    '''
    
    '''
    def create_bottomBar(self):
        # Rahmen 
        self.bottomBar_frame = ctk.CTkFrame(self.window_frame, height=30, fg_color='#1a1933')
        self.bottomBar_frame.grid(row=1, column=0, columnspan=3, sticky='ew')
        self.bottomBar_frame.grid_propagate(False)

if __name__ == '__main__':
    gui = GUI()
    gui.mainloop()
```
